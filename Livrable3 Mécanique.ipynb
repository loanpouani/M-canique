```python
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
import math
import sys


#  TABLEAU DES MODÈLES
#  modele[nom, masse, acceleration moyenne, longueur, largeur, hauteur, Cx, Cz, mu]

modele = [
    ["Dodge Charger R/T, 1970", 1760, 5.1, 5.28, 1.95, 1.35, 0.38, 0.3, 0.1],
    ["Toyota Supra Mark IV, 1994", 1615, 5.0, 4.51, 1.81, 1.27, 0.29, 0.3, 0.1],
    ["Chevrolet Yenko Camaro 1969", 1498, 5.3, 4.72, 1.88, 1.30, 0.35, 0.3, 0.1],
    ["Mazda RX-7 FD", 1385, 5.2, 4.3, 1.75, 1.23, 1.28, 0.3, 0.1],
    ["Nissan Skyline GTR-R34, 1999", 1540, 5.8, 4.6, 1.79, 1.36, 0.34, 0.3, 0.1],
    ["Mitsubishi Lancer Evolution VII", 1600, 5.0, 4.51, 1.81, 1.48, 0.28, 0.3, 0.1]
]

#  CHOIX DU MODÈLE
print("Choisissez votre modèle de voiture :")
for i, m_info in enumerate(modele):
    print(f"{i+1} : {m_info[0]}")
chiffre = int(input("Entrez un chiffre entre 1 et 6 : "))
while chiffre < 1 or chiffre > 6:
    chiffre = int(input("Réponse invalide. Entrez un chiffre entre 1 et 6 : "))
q = chiffre - 1
print("Vous avez choisi :", modele[q][0])

#  CHOIX DE L'ACCESSOIRE
print("\nAccessoires : 0 = Aucun, 1 = NOS, 2 = Ailerons, 3 = Jupe avant")
choix_accessoire = int(input("Entrez votre choix : "))
while choix_accessoire not in (0, 1, 2, 3):
    choix_accessoire = int(input("Réponse invalide. Choisissez 0, 1, 2 ou 3 : "))

# Si NOS, demander sur quelle partie
nitro_part = 0
nitro_multiplier = 1.0
if choix_accessoire == 1:
    nitro_part = int(input("Sur quelle partie activer le NOS ? 1(Pente), 2(Looping), 3(Saut), 4(Ligne Droite) : "))
    while nitro_part not in (1, 2, 3, 4):
        nitro_part = int(input("Réponse invalide. Sur quelle partie activer le NOS ? 1-4 : "))
    nitro_multiplier = 1.3

#  HAUTEUR ET DISTANCE DE LA PENTE/SAUT
h_pente = float(input("Hauteur verticale de la pente [m] : "))
L_target = 9.0 # Distance cible du saut (pour Programme 3)
H_target = 0.0 # Hauteur cible du saut (pour Programme 3)

#  INITIALISATION DES VARIABLES GLOBALES ET DE TRANSITION

# Variables de transition (vitesses)
v_fin_pente = 0.0
v_fin_looping = 0.0
v_fin_saut = 0.0
v_fin_ld = 0.0

# Variables de temps
temps_pente = 0.0
temps_looping = 0.0
temps_saut = 0.0
temps_ligne_droite = 0.0
temps_total_parcours = 0.0

# Variables à partir du modèle
m = modele[q][1]
atr_base = modele[q][2]  # accélération moyenne
longueur = modele[q][3]
w_vehicule = modele[q][4]
h_vehicule = modele[q][5]
Cx = modele[q][6]
Cz = modele[q][7]
mu = modele[q][8]

# Application accessoire (modification des variables globales)
if choix_accessoire == 2:  # Ailerons
    Cz *= 1.10
    m += 30
elif choix_accessoire == 3:  # Jupe avant
    Cx *= 0.95
    m += 15

Sx = longueur * h_vehicule # Surface frontale (pour Ligne droite et saut)
Sz = longueur * w_vehicule # Surface portante (pour saut)
rho = 1.225
g = 9.81
L_distance = 31.0  # longueur standard de la pente


#  PROGRAMME 1 : PENTE (AVEC AFFICHAGE ET GRAPHIQUES)
def programme_pente():
    global v_fin_pente, temps_pente, temps_total_parcours
    
    print("\n--- MODULE 1 : PENTE ---")
    
    x0 = 0
    v0 = 0
    S_frontale = w_vehicule * h_vehicule  # Surface frontale
    atr = atr_base
    if choix_accessoire == 1 and nitro_part == 1:
        atr *= nitro_multiplier

    # CALCULS GEOMETRIQUES
    alpha = np.arcsin(h_pente / L_distance)
    alpha_deg = np.rad2deg(alpha)
    sin_alpha = np.sin(alpha)
    cos_alpha = np.cos(alpha)
    print(f"Angle calculé : α = {alpha_deg:.2f}°")
    
    dt = 0.01
    t = 0.0
    x = x0
    v = v0
    
    # Vecteurs pour les graphes
    T = []
    X = []
    V = []
    A = []
    a = 0.0 # Initialisation de l'accélération
    
    # TABLEAU COMPLET
    print("\nTableau t | x(t) :")
    print(" t [s]       x(t) [m]")
    print("--------------------------")
    
    # Simulation (méthode d'Euler du programme original)
    while x < L_distance:
        
        # Calcul traînée
        drag = (0.5 * rho * S_frontale * Cx * v * abs(v)) / m

        # Accélération instantanée
        a = atr + g*sin_alpha - mu*g*cos_alpha - drag
        
        # Stockage pour graphes
        T.append(t)
        X.append(x)
        V.append(v)
        A.append(a)

        # affichage intermédiaire
        # Affichage toutes les 25 étapes pour éviter un défilement trop rapide
        if len(T) % 25 == 0:
            print(f"{t:7.3f}     {x:10.4f}")

        # incrémentation
        v = v + a * dt
        x = x + v * dt
        t += dt

    # Affiche la dernière ligne, même si x > L_distance
    print(f"{t:7.3f}     {x:10.4f}    <-- Atteint la distance cible")
    
    # Mise à jour des variables globales
    temps_pente = t
    v_fin_pente = v
    temps_total_parcours += t

    # AFFICHAGE FINAL
    print("\n--------------------------------------")
    print(f"Atteint {L_distance} m à t = {t:.3f} s")
    print(f"Vitesse finale  = {v:.3f} m/s")
    print(f"Accélération    = {a:.3f} m/s²")
    print("--------------------------------------\n")

    # GRAPHIQUES 
    plt.figure(figsize=(11,9))
    
    plt.subplot(3,1,1)
    plt.plot(T, X)
    plt.axhline(L_distance, color="grey", linestyle="--")
    plt.ylabel("Position [m]")
    plt.title(f"Pente : Position x(t) pour le modèle {modele[q][0]}")
    plt.grid(True)

    plt.subplot(3,1,2)
    plt.plot(T, V)
    plt.ylabel("Vitesse [m/s]")
    plt.title("Pente : Vitesse v(t)")
    plt.grid(True)

    plt.subplot(3,1,3)
    plt.plot(T, A)
    plt.xlabel("Temps [s]")
    plt.ylabel("Accélération [m/s²]")
    plt.title("Pente : Accélération a(t)")
    plt.grid(True)

    plt.tight_layout()
    plt.show()


   #  PROGRAMME 2 : LOOPING
def programme_looping(): #On définini une fonction qui englobe toute la simulation du looping
    global v_fin_looping, temps_looping, temps_total_parcours #Nos variables globales 
    print("\n--- MODULE 2 : LOOPING ---")  # On Affiche le titre de la partie dans la console
 
    R = 6  # rayon du looping
    v_init = v_fin_pente # On récupère la vitesse obtenue après la pente
    # On verifie que la vitesse est suffisante pour passer le looping
    if v_init < 10:
        print("Vitesse initiale trop faible pour le Looping. Simulation annulée.")
        v_fin_looping = 0.0 #Si c'est =0  = Résultat nul
        temps_looping = 9999.0 # Temps pour signaler l’échec
        return # On arrête la fonction
    a_moy = atr_base #On déclare l'accélération moyenne de base
    if choix_accessoire == 1 and nitro_part == 2: # Si l'accessoire nitro est activé
        a_moy *= nitro_multiplier # On multiplie l’accélération par le boost nitro
 
    Surface = w_vehicule * h_vehicule # On déclare la Surface frontale
    theta0 = 0 # On déclare theta l'Angle initial 0 
    thetadot0 = v_init / R #On déclare la Vitesse angulaire initiale
    conds_init = [theta0, thetadot0] #On déclare nos conditions initales avec theta et thetapoint
 
    # Déclaration de l'équation différentielle
    def eq_loop(S, t): 
        theta, thetadot = S # On récupère l'angle et la vitesse angulaire
        thetaddot = (a_moy/R # L'accélération moteur 
                     - (g/R)*np.cos(theta) #gravité 
                     + (mu*g/R)*np.sin(theta) #frottement 
                     - thetadot**2 * ((rho*Surface*Cx*R)/(2*m) + mu)) #trainée + frottement
        return [thetadot, thetaddot] # Retourne vitesse angulaire + accélération angulaire
 
    # Résolution et recherche du temps pour 2*pi
    t_sim = np.linspace(0, 5, 1000) # Simuler sur 5s avec 1000 pas
    sol = odeint(eq_loop, conds_init, t_sim)  # Solution de l’équation différentielle
 
    v = R * sol[:, 1] # Vitesse linéaire v = R * vitesse angulaire
    theta = sol[:, 0] # Position angulaire en fonction du temps
 
    
    # Vérification de la réaction normale (Point clé du Looping)
    N = m * ( (v**2)/R + g*np.cos(theta) )  # On déclare l'équation de la réaction normale

    if np.min(N) <= 0: #Ensuite on vérifie qu'elle est bien supérieur à 0 car si elle serait nulle cela voudrait dire que les roues ne touche pas le sol et que la voiture tombe
        idx_fail = np.where(N <= 0)[0] # On trouve le moment exact où ça arrive
        if idx_fail.size > 0:
             print(f"La réaction normale est nulle à t={t_sim[idx_fail[0]]:.3f}s. La voiture tombe du looping.") #On affiche un message si la reaction normal est nulle et on précise a quelle moment
             temps_looping = 9999.0
             v_fin_looping = 0.0 
             return
  # On cherche quand theta egale 2π (un tour complet)
    idx_end = np.where(theta >= 2 * np.pi)[0]
 
    if idx_end.size > 0: # Si un tour complet est réussi
        idx = idx_end[0] # Premier instant où theta = 2π
        temps_looping = t_sim[idx] # Temps pour faire un tour
        v_fin_looping = v[idx] # Vitesse de sortie du looping
        temps_total_parcours += temps_looping  # On ajoute au temps total du circuit
        print(f"LOOPING terminé en {temps_looping:.3f} s. Vitesse de sortie : {v_fin_looping:.3f} m/s") #On affiche le résultat
    else:
        temps_looping = 9999.0
        v_fin_looping = 0.0
        print("ÉCHEC : La voiture n'a pas fait un tour complet du looping.") #On affiche un message d'erreur/echec si la voiture ne fait pas le looping
 
 
    # Représentation graphique 
    plt.figure(figsize=(10,5)) #On créer un graphique
    plt.plot(t_sim, v, linewidth=2) #On choisi notre épaisseur
    plt.xlabel("Temps (s)")#Nom de l'axe X
    plt.ylabel("Vitesse (m/s)") #Nom de l'axe Y
    plt.title(f"Looping : Vitesse v(t) pour le modèle {modele[q][0]}") #Titre du schéma
    plt.grid(True) #On met des grilles pour que la courbe et les résultats soit mieux visible
    plt.show() #On affiche le graphique

   

#  PROGRAMME 3 : SAUT

def programme_saut():
    global v_fin_saut, temps_saut, temps_total_parcours
    
    print("\n--- MODULE 3 : SAUT / TRAJECTOIRE ---")

    if v_fin_looping <= 0.0:
        print("Vitesse initiale du Saut insuffisante. Simulation annulée.")
        temps_saut = 9999.0
        v_fin_saut = 0.0
        return
        
    # Positions et vitesses initiales
    x0 = 0
    y0 = h_vehicule 
    v0x = v_fin_looping # Vitesse à la fin du looping (globale)
    v0y = 0

    # Coefficients aérodynamiques
    Cx_eff = Cx
    Cz_eff = Cz
    if choix_accessoire == 1 and nitro_part == 3:
        v0x *= nitro_multiplier # Application du NOS
        
    # Surfaces
    Sx_saut = longueur * h_vehicule
    Sz_saut = longueur * w_vehicule

    # Paramètres aérodynamiques
    kx = 0.5 * rho * Cx_eff * Sx_saut
    kz = 0.5 * rho * Cz_eff * Sz_saut

    # Conditions initiales pour odeint
    S_init = [x0, y0, v0x, v0y]
    params_saut = (kx, kz, m, g) # Paramètres pour odeint

    # Définition de l'équation différentielle
    def Sp(S, t, kx, kz, m, g):
        vx = S[2]
        vy = S[3]
        V = np.sqrt(vx**2 + vy**2) + 1e-12 
        
        ax = -kx/m * V * vx
        ay = -g - kz/m * V * vy
        
        return [vx, vy, ax, ay]

    # Résolution de l'équation différentielle
    t_sim = np.linspace(0, 3, 500) # Simuler sur 3s
    S = odeint(Sp, S_init, t_sim, args=params_saut)

    x_saut = S[:, 0]
    y_saut = S[:, 1]
    
    # Trouver le temps où x atteint la cible
    indices_x = np.where(x_saut >= L_target)[0]

    if indices_x.size > 0:
        idx_target = indices_x[0]
        temps_saut = t_sim[idx_target]
        y_atterrissage = y_saut[idx_target]
        
        v_x_fin = S[idx_target, 2]
        v_y_fin = S[idx_target, 3]
        v_fin_saut = np.sqrt(v_x_fin**2 + v_y_fin**2)
        
        temps_total_parcours += temps_saut
        
        print(f"SAUT: Atteint {L_target}m en {temps_saut:.3f} s. Atterrissage y={y_atterrissage:.3f} m.")
        if y_atterrissage < H_target:
            print("ÉCHEC: La voiture est tombée dans le ravin.")
        
    else:
        temps_saut = 9999.0
        v_fin_saut = 0.0
        print(f"ÉCHEC: La voiture n'a pas atteint la distance cible de {L_target} m.")
        
    # Affichage de la trajectoire 
    plt.figure(figsize=(8,5))
    plt.plot(x_saut, y_saut, color="blue", lw=2, label="Trajectoire")
    plt.scatter(L_target, H_target, color='red', label="Cible")
    plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)  # sol
    plt.xlabel("x (m)")
    plt.ylabel("y (m)")
    plt.title(f"Saut : Trajectoire pour le modèle {modele[q][0]}")
    plt.legend()
    plt.grid(True)
    plt.show()

# =========================
#  PROGRAMME 4 : LIGNE DROITE
# =========================
def programme_ligne_droite():
    global v_fin_ld, temps_ligne_droite, temps_total_parcours

    print("\n--- MODULE 4 : LIGNE DROITE ---")
    
    dfinale = 10.0 # Distance cible
    
    am4 = atr_base
    if choix_accessoire == 1 and nitro_part == 4: am4 *= nitro_multiplier
    
    params = (rho, Sx, Cx, m, am4, mu, g)
    
    # Définition de l'équation différentielle
    def dvx_dt_simple(vx, t, params):
        rho_p, Sx_p, Cx_p, m_p, am_p, mu_p, g_p = params
        # Équation: dv/dt = (F_traction - F_trainée - F_frottement) / m
        return (-0.5 * rho_p * Sx_p * Cx_p * vx**2) / m_p + am_p - mu_p * g_p
        
    t_sim = np.linspace(0, 10, 1000)
    v0x = v_fin_saut # Vitesse initiale (sortie du saut) 
    v_ld = odeint(dvx_dt_simple, v0x, t_sim, args=(params,)).flatten()
    
    # Calcul de la position (intégration numérique de la vitesse)
    dt_ld = t_sim[1] - t_sim[0]
    x_ld = np.cumsum(v_ld) * dt_ld
    
    # Trouver le temps où x atteint la distance cible
    indices_x_ld = np.where(x_ld >= dfinale)[0]

    if indices_x_ld.size > 0:
        idx_ld_end = indices_x_ld[0]
        v_fin_ld = v_ld[idx_ld_end]
        temps_ligne_droite = t_sim[idx_ld_end]
        temps_total_parcours += temps_ligne_droite
        print(f"LIGNE DROITE terminée en {temps_ligne_droite:.3f} s. Vitesse finale : {v_fin_ld:.3f} m/s")
    else:
        temps_ligne_droite = 9999.0
        v_fin_ld = 0.0
        print(f"ÉCHEC : La voiture n'a pas atteint la distance cible de {dfinale} m.")

    # Affichage du graphique de vitesse 
    plt.figure(figsize=(10,6))
    plt.plot(t_sim,v_ld,label='Vitesse (m/s)')
    plt.title(f"Ligne Droite : Évolution de la vitesse en fonction du temps pour le modèle {modele[q][0]}")
    plt.xlabel("Temps (s)"); plt.ylabel("Vitesse (m/s)"); plt.grid(True); plt.legend(); plt.show()


# =========================
#  EXECUTION AUTOMATIQUE DES 4 PROGRAMMES
# =========================
programme_pente()
programme_looping()
programme_saut()
programme_ligne_droite()

# =========================
#  RÉSUMÉ FINAL
# =========================
print("\n" + "="*50)
print(f"RÉSUMÉ FINAL DU PARCOURS pour le modèle {modele[q][0]}")
print("="*50)
print(f"1. PENTE : {temps_pente:.3f} s (Sortie: {v_fin_pente:.2f} m/s)")
print(f"2. LOOPING : {temps_looping:.3f} s (Sortie: {v_fin_looping:.2f} m/s)")
print(f"3. SAUT : {temps_saut:.3f} s (Sortie: {v_fin_saut:.2f} m/s)")
print(f"4. LIGNE DROITE : {temps_ligne_droite:.3f} s (Sortie: {v_fin_ld:.2f} m/s)")
print("--------------------------------------------------")

if 9999.0 in [temps_pente, temps_looping, temps_saut, temps_ligne_droite]:
    print("LE PARCOURS N'A PAS ÉTÉ COMPLÉTÉ.")
else:
    print(f"TEMPS TOTAL DU PARCOURS : {temps_total_parcours:.3f} secondes")
print("="*50)
```

Une fois les caractéristiques de la voiture ajustées, le code calcule et définit les constantes nécessaires à tous les modules de la simulation : Sx et Sz sont les Surfaces d'impact de la voiture. Le code calcule la surface frontale (Sx = longueur * h_vehicule) et la surface portante (Sz = longueur * w_vehicule`) pour savoir quelle quantité d'air la voiture déplace. rho = 1.225 est la masse volumique de l'air (en kg/m^3), nécessaire pour calculer la traînée aérodynamique. g = 9.81 est la constante d'accélération due à la gravité (en m/s^2). L_distance = 31.0 définit la longueur de la Pente (en mètres).

```python
def programme_pente():
    print("\n MODULE 1 : PENTE")
    
    x0 = 0
    v0 = 0
    S = h_vehicule * w_vehicule  # Surface frontale
    atr = atr_base
    # appliquer NOS sur la partie choisie
    if choix_accessoire == 1 and nitro_part == 1:
       atr *= nitro_multiplier
    
    
    
    alpha = np.arcsin(h_pente / L_distance)
    alpha_deg = np.rad2deg(alpha)
    sin_alpha = np.sin(alpha)
    cos_alpha = np.cos(alpha)
    print(f"Angle calculé : α = {alpha_deg:.2f}°")

    dt = 0.01             # pas de temps
    t = 0.0               # temps actuel
    x = x0                # position
    v = v0                # vitesse

    # Vecteurs pour les graphes
    T = []
    X = []
    V = []
    A = []
```

La ligne def programme_pente(): donne le nom de "programme pente" à toutes les instructions qui suivent, créant ainsi une partie dédiée au calcul de ce morceau de la course. Les lignes qui suivent affichent un titre et préparent la voiture pour le départ : x0 = 0 (elle est à 0 mètre) et v0 = 0 (elle est à 0 m/s, donc à l'arrêt). L'ordinateur calcule ensuite la Surface frontale (S) de la voiture en multipliant sa hauteur par sa largeur. Elle va nous servir à déterminé à quel point la voiture sera freinée par l'air (la traînée). Le code définie l'accélération normale du moteur (atr_base). Enfin, il vérifie si vous avez prévu d'utiliser le NOS (choix_accessoire == 1) précisément sur la descente (nitro_part == 1). Si oui, il applique immédiatement le boost en multipliant l'accélération par 1.3, donnant à la voiture sa puissance maximale au moment de l'élan.

On determine l'inclinaison exacte de la piste :alpha = np.arcsin(h_pente / L_distance) La Pente est considérée comme un triangle rectangle où h_pente est la hauteur verticale et L_distance est l'hypoténuse (la longueur de la piste). Cette ligne utilise la fonction arc-sinus (la réciproque du sinus) pour trouver l'angle alpha de la pente, en radians. alpha_deg = np.rad2deg(alpha) : Convertit cet angle alpha des radians en degrés sin_alpha = np.sin(alpha) et cos_alpha = np.cos(alpha) : Calcule le Sinus et le Cosinus de cet angle. Ces valeurs sont indispensables dans les équations de physique pour décomposer la force de gravité (G) en une force qui tire la voiture vers le bas de la pente (G sin(alpha)) et une force perpendiculaire à la piste (Gcos(alpha)). Print(f"Angle calculé : α = {alpha_deg:.2f}°") : Affiche l'angle que l'ordinateur a calculé.2. Initialisation des Variables pour le au cours de temps. Cette partie prépare le système pour enregistrer le déroulement de la simulation, étape par étape :dt = 0.01 : Définit le pas de temps (dt). C'est l'intervalle de temps entre chaque calcul. Ici, le programme fera un nouveau calcul toutes les 0,01 secondes. Plus ce chiffre est petit, plus la simulation est précise. t = 0.0, x = x0, v = v0 : Confirme le démarrage à t=0 seconde, à la position de départ x=x_0 (qui est 0), et à la vitesse v=v_0 (qui est 0).
